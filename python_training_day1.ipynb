```python
#Accept 5 paper marks, rollno, name; and calculate total & percentage if percenatge >=85, student can take admission else can't take addmission  using class & object constructor

class Student:
    def __init__(self, rollno, name, marks):
        self.rollno = rollno
        self.name = name
        self.marks = marks

    def calculate(self):
        total = sum(self.marks)
        percentage = total / 5

        print("\nRoll No:", self.rollno)
        print("Name:", self.name)
        print("Total Marks:", total)
        print("Percentage:", percentage)

        if percentage >= 85:
            print("Eligible for Admission")
        else:
            print("Not Eligible for Admission")


rollno = int(input("Enter Roll No: "))
name = input("Enter Name: ")

marks = []
for i in range(5):
    m = float(input("Enter paper marks: "))
    marks.append(m)

s = Student(rollno, name, marks)
s.calculate()

```

```python
#single level inheritance
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

obj=B()
obj.showA()
obj.showB()
```

```python
#multi level inheritance
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

class C(B):
  def showC(self):
    print("showC function")

obj=C()
obj.showA()
obj.showB()
obj.showC
```

```python
#multi level inheritance
class A:
  def showA(self):
    print("showA function")

class B:
  def showB(self):
    print("showB function")

class C(A,B):
  def showC(self):
    print("showC function")

obj=C()
obj.showA()
obj.showB()
obj.showC
```

```python
# Hybrid inheritance

class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")

class D(B, C):
    def showD(self):
        print("showD function")


obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
```

```python
# Hierarchical inheritance

class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")

obj1 = B()
obj1.showA()
obj1.showB()

obj2 = C()
obj2.showA()
obj2.showC()
```
