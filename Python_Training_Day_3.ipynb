```python
#NUMPY
import numpy as np

students = np.array([
    [85, 90, 78],  # Student 1
    [60, 55, 70],  # Student 2
    [95, 92, 88],  # Student 3
    [40, 50, 45],  # Student 4
    [72, 68, 75],   # Student 5
])

students
```

```python
avg_marks=np.mean(students,axis=1)
print("Average marks:",avg_marks)
```

```python
topper=np.argmax(avg_marks)
print("Topper Student:",topper)
```

```python
failed_student=np.where(avg_marks<50)
print("Failed Students:",failed_student[0]+1)
```

```python
#performance report
for i, avg in enumerate(avg_marks):
    status = "Pass" if avg >= 50 else "Fail"
    print(f"Student {i+1}: Avg = {avg:.2f} | {status}")
```

```python
avg_marks = np.mean(students, axis=0)
for i, avg in enumerate(avg_marks):
    status = "Pass" if avg >= 50 else "Fail"
    print(f"Subject {i+1}: Avg = {avg:.2f} | {status}")
```

```python
data = {
    'Name': ['A', 'B', 'C', 'D', 'E'],
    'Age': [78, 67, 98, 56, 88],
    'Science': [87, 89, 98, 67, 92],
    'English': [88, 83, 86, 89, 98]
}
df = pd.DataFrame(data)
print(df)
```

```python
marks = df[["Science", "English"]].values
average_marks = np.mean(marks, axis=1)
df["Average"] = average_marks
print(df)
```

```python
def grade(avg):
    if avg >= 90:
        return 'A'
    elif avg >= 75:
        return 'B'
    elif avg >= 60:
        return 'C'
    else:
        return 'D'

# Apply grading function
df['Grade'] = df['Average'].apply(grade)

print(df)
```

```python
plt.figure()
plt.bar(df["Name"], df["Average"])
plt.xlabel('Students')
plt.ylabel('Average Marks')
plt.title('Average Marks of Students')
plt.show()
```
