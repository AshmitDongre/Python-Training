```python
#Program to print the area of rectangle by creating a class named 'Area' taking the values of its length and breadth as parameters of its constructor and having a method named 'returnarea' which returns the area of the rectangle . Lenth and breadth of rectangle are entered through keyboard
class Area:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def returnarea(self):
        return self.length * self.breadth


length = float(input("Enter length: "))
breadth = float(input("Enter breadth: "))

obj = Area(length, breadth)

print("Area of Rectangle:", obj.returnarea())
```

```python
class Area:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def returnArea(self):
        area = self.length * self.breadth
        print("Area of Rectangle:", area)


length = int(input("Enter length: "))
breadth = int(input("Enter breadth: "))

obj = Area(length, breadth)
obj.returnArea()
```

```python
#from abc import
from abc import ABC, abstractmethod

class Test(ABC):
    @abstractmethod
    def m1(self):
        pass

    @abstractmethod
    def m2(self):
        pass


class MyPython(Test):
    def m1(self):
        print("m1 implementation")

    def m2(self):
        print("m2 implementation")


obj = MyPython()
obj.m1()
obj.m2()
```

```python
# Inside the abstract class we can declare the constructor.
# Abstract class constructor is executed but object is not created.

from abc import ABC, abstractmethod

class Test(ABC):
    def __init__(self):
        print("Constructor is called...")

    @abstractmethod
    def m1(self):
        pass


class MyPython(Test):
    def m1(self):
        print("m1 Method Implementation")


m = MyPython()
m.m1()
```

```python
#we can use seek() function to move cursor(file pointer) to specifies location

data = "Learning python is very easy from Ashish sir"

f = open("myfile.txt", 'w')
f.write(data)
f.close()

with open("myfile.txt", 'r+') as f:
    fileData = f.read()
    print(fileData)

    print("Current Cursor Position is", f.tell())

    f.seek(15)
    print("Current Cursor Position is", f.tell())

    f.write("-JAVA-")

    f.seek(0)
    fileData = f.read()
    print(fileData)
```

```python
#WAP to print number of lines , words, characters present in file
import os

fname = input("Enter file name : ")
if os.path.isfile(fname):
    print("File exist : ", fname)

    lcount = wcount = ccount = 0
    # Using 'with open' is best practice as it automatically closes the file
    with open(fname, 'r') as f:
        for line in f:
            lcount = lcount + 1
            ccount = ccount + len(line)
            words = line.split()
            wcount = wcount + len(words)

    print("Total lines : ", lcount)
    print("Total words : ", wcount)
    print("Total chracters : ", ccount)
```
